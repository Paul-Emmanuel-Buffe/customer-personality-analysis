### 1.2 création de la table + chargement du CSV (si table vide)

```python
with psycopg2.connect(**CONN_PARAMS) as conn:
    with conn.cursor() as cur:
        cur.execute("""
            CREATE TABLE IF NOT EXISTS clients (
                id INTEGER PRIMARY KEY, year_birth INTEGER, education VARCHAR(50),
                marital_status VARCHAR(50), income NUMERIC, kidhome INTEGER,
                teenhome INTEGER, dt_customer DATE, recency INTEGER,
                mntwines NUMERIC, mntfruits NUMERIC, mntmeatproducts NUMERIC,
                mntfishproducts NUMERIC, mntsweetproducts NUMERIC, mntgoldprods NUMERIC,
                numdealspurchases INTEGER, numwebpurchases INTEGER,
                numcatalogpurchases INTEGER, numstorepurchases INTEGER,
                numwebvisitsmonth INTEGER, acceptedcmp3 INTEGER, acceptedcmp4 INTEGER,
                acceptedcmp5 INTEGER, acceptedcmp1 INTEGER, acceptedcmp2 INTEGER,
                complain INTEGER, response INTEGER
            );
        """)
        conn.commit()

        cur.execute("SELECT COUNT(*) FROM clients;")
        nb_lignes_en_base = cur.fetchone()[0]

        if nb_lignes_en_base == 0:
            print("Table vide -> chargement du CSV...")
            df_brut = pd.read_csv(CHEMIN_CSV, sep="\t")
            df_brut["Dt_Customer"] = pd.to_datetime(df_brut["Dt_Customer"], format="%d-%m-%Y")

            colonnes = [
                "ID", "Year_Birth", "Education", "Marital_Status", "Income", "Kidhome",
                "Teenhome", "Dt_Customer", "Recency", "MntWines", "MntFruits",
                "MntMeatProducts", "MntFishProducts", "MntSweetProducts", "MntGoldProds",
                "NumDealsPurchases", "NumWebPurchases", "NumCatalogPurchases",
                "NumStorePurchases", "NumWebVisitsMonth", "AcceptedCmp3", "AcceptedCmp4",
                "AcceptedCmp5", "AcceptedCmp1", "AcceptedCmp2", "Complain", "Response",
            ]
            df_brut = df_brut[colonnes]
            valeurs = [tuple(row) for row in df_brut.to_numpy()]
            execute_values(
                cur,
                f"INSERT INTO clients ({', '.join(df_brut.columns.str.lower())}) VALUES %s",
                valeurs
            )
            conn.commit()
            print(f"{len(df_brut)} lignes insérées.")
        else:
            print(f"Table déjà peuplée ({nb_lignes_en_base} lignes) -> pas de rechargement.")
```

## 2. Dictionnaire des données (Data Dictionary)

### 1. People (Profil Client)
* **ID :** Identifiant unique du client.
* **Year_Birth :** Année de naissance du client.
* **Education :** Niveau d'études du client.
* **Marital_Status :** État civil du client.
* **Income :** Revenu annuel brut du foyer du client.
* **Kidhome :** Nombre d'enfants en bas âge dans le foyer.
* **Teenhome :** Nombre d'adolescents dans le foyer.
* **Dt_Customer :** Date d'enregistrement du client auprès de l'entreprise.
* **Recency :** Nombre de jours écoulés depuis le dernier achat.
* **Complain :** 1 si le client s'est plaint au cours des 2 dernières années, 0 sinon.

### 2. Products (Dépenses par Catégorie)
* **MntWines :** Montant dépensé en vin au cours des 2 dernières années.
* **MntFruits :** Montant dépensé en fruits au cours des 2 dernières années.
* **MntMeatProducts :** Montant dépensé en viande au cours des 2 dernières années.
* **MntFishProducts :** Montant dépensé en poisson au cours des 2 dernières années.
* **MntSweetProducts :** Montant dépensé en confiseries au cours des 2 dernières années.
* **MntGoldProds :** Montant dépensé en produits de luxe/or au cours des 2 dernières années.

### 3. Promotion (Comportement Campagnes)
* **NumDealsPurchases :** Nombre d'achats effectués avec une réduction.
* **AcceptedCmp1 :** 1 si le client a accepté l'offre de la 1ère campagne, 0 sinon.
* **AcceptedCmp2 :** 1 si le client a accepté l'offre de la 2ème campagne, 0 sinon.

* **AcceptedCmp4 :** 1 si le client a accepté l'offre de la 4ème campagne, 0 sinon.
* **AcceptedCmp5 :** 1 si le client a accepté l'offre de la 5ème campagne, 0 sinon.
* **Response :** 1 si le client a accepté l'offre de la dernière campagne marketing, 0 sinon.

### 4. Place (Canaux d'Achat)

### 2.1. Feature Engineering : Calcul de l'Âge et de l'Ancienneté à l'Été 2014

> **Le Scénario Métier :** "Nous nous plaçons dans la posture d'un data analyst au **30 juin 2014**, au lendemain même de la dernière extraction de données. Nous étudions la base client d'une enseigne récente lancée fin juillet 2012, nous offrant ainsi un recul exact de **23 mois** sur son démarrage et sur le comportement d'achat de ses premières cohortes de clients."

Nous définissons cette date du 30/06/2014 comme notre point d'ancrage historique pour recalculer nos variables temporelles :
* **Age :** L'âge du client au moment de l'étude (en 2014).
* **Anciennete_Jours :** Le nombre de jours réels écoulés entre l'inscription du client (`Dt_Customer`) et le 30 juin 2014.

*Note : Les colonnes d'origine `Year_Birth` et `Dt_Customer` sont supprimées après traitement.*

```python
DATE_REFERENCE = pd.to_datetime('2014-06-30')

valeurs_age = DATE_REFERENCE.year - df['year_birth']

df['dt_customer'] = pd.to_datetime(df['dt_customer'], format='%d-%m-%Y')
valeurs_anciennete = (DATE_REFERENCE - df['dt_customer']).dt.days

df.drop(columns=['year_birth', 'dt_customer'], inplace=True)

# Insertion de la nouvelle colonne à l'idex 1
df.insert(1, 'age', valeurs_age)
df.insert(2, 'ancienne_jours', valeurs_anciennete)
df.head()
```

```python
# Configuration de l'affichage (1 ligne, 2 graphiques)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Boxplot de l'Âge (Calculé précédemment sur DATE_REFERENCE)
sns.boxplot(data=df, x='age', ax=axes[0], color='skyblue')
axes[0].set_title("Visualisation de l'Outlier d'Âge")
axes[0].set_xlabel("Âge (en 2014)")

# 2. Boxplot des Revenus (Income)
sns.boxplot(data=df, x='income', ax=axes[1], color='salmon')
axes[1].set_title("Visualisation de l'Outlier de Revenu")
axes[1].set_xlabel("Revenu Annuel")

plt.tight_layout()
plt.show()
```

**Synthèse:**
L'analyse des statistiques globales et des visualisations (Boxplots) sur les 2 240 clients révèle des anomalies nettes et oriente nos choix de nettoyage pour garantir la stabilité du clustering :

* **Âge :** Le minimum de 18 ans valide la conformité réglementaire de la base. En revanche, le graphique isole clairement trois profils aberrants au-delà de 110 ans. Nous appliquons un filtrage strict à 100 ans (`df['Age'] <= 100`) : cela supprime ces anomalies sans perte d'information utile, un choix également cohérent sur le plan métier car ces profils ne sont généralement plus les décideurs d'achat principaux.
* **Revenus (Income) :** La colonne présente 24 valeurs manquantes (qui seront imputées par la médiane) ainsi qu'une erreur de saisie évidente à 666 666 €. Pour la modélisation, nous choisissons un filtrage strict au seuil de 120 000 € (`df['Income'] <= 120000`). Bien que les hauts revenus jusqu'à 200 000 € soient économiquement plausibles, leur écart extrême par rapport à la masse de la clientèle fausserait mathématiquement les distances euclidiennes du K-Means, risquant de monopoliser un cluster entier au détriment de la précision globale.
* **Foyers :** Plus de la moitié des clients n'ont aucun enfant ni adolescent à charge. Les profils avec enfants à charge existent (jusqu'à 4 au total par cumul) mais restent minoritaires.
* **Récence :** La répartition est parfaitement uniforme de 0 à 99 jours. Cet étalement artificiel prouve que les clients inactifs ont été supprimés à la source. Tous les profils étant actifs, cette variable n'aidera pas à séparer les clients lors du clustering.

### 2.3 Exploration des Variables Catégorielles

```python
# Analyse des fréquences pour les variables qualitatives majeures
variables_cat = ['education', 'marital_status']

for col in variables_cat:
    print(f"--- Répartition de la variable : {col} ---")
    print(df[col].value_counts(dropna=False))
    print(f"Nombre de modalités uniques : {df[col].nunique()}")
    print("-" * 40 + "\n")
```

**Synthèse:**
L'examen des variables qualitatives (`Education` et `Marital_Status`) met en évidence des disparités d'effectifs et des incohérences de saisie. Afin d'optimiser l'Analyse à Facteurs Multiples (AFM) et d'éviter l'instabilité des modèles, les arbitrages suivants ont été retenus :

* **Éducation :** * **Fusion de `2n Cycle` et `Master` :** Ces deux modalités désignent réglementairement le même niveau universitaire (Bac+5). Leur regroupement crée un bloc statistiquement plus robuste.
    * **Maintien de `Graduation` (Licence) et `PhD` (Doctorat) séparés :** Ces diplômes correspondent à des tranches d'âge et des catégories socio-économiques distinctes, indispensables pour qualifier le comportement d'achat.
    * **Maintien de `Basic` isolé :** Malgré son faible effectif (54 clients), ce profil non-diplômé du supérieur présente un comportement de consommation potentiellement très différent à l'épicerie qu'il convient de ne pas noyer dans la masse.

* **Statut Marital :**
    * **Suppression de `Absurd` :** Cette modalité constitue un bruit informatique pur sans aucune valeur métier. Représentant une part infime du dataset, elle est éliminée.
    * **Intégration de `Alone` et `YOLO` dans `Single` :** `Alone` est un synonyme parfait de célibataire. Pour `YOLO`, l'isolement de seulement deux individus aurait créé une dimension mathématique vide et instable lors de l'encodage pour l'AFM. Les basculer dans `Single` permet de conserver leur historique d'achat sans distordre ce grand groupe.
    * **Maintien de `Widow` (Veuf/Veuve) isolé :** Avec 77 clients, cet effectif est suffisant pour être exploité. Fusionner ces profils avec les `Single` (généralement plus jeunes) détruirait un signal marketing "Senior" très fort, caractérisé par des habitudes d'achat spécifiques (portions, rayons dédiés).

## 3. Matrice de corrélation

```python
# 1. Sélection des colonnes numériques (tout en minuscules)
colonnes_analyse = [
    'age', 
    'income', 
    'kidhome', 
    'teenhome', 
    'recency',
    'mntwines', 
    'mntmeatproducts', 
    'mntfruits', 
    'mntgoldprods',
    'numwebpurchases', 
    'numstorepurchases', 
    'numdealspurchases'
]

# 2. Calcul de la matrice de corrélation
matrice_corr = df[colonnes_analyse].corr()

# 3. Affichage sous forme de Heatmap thermique
plt.figure(figsize=(12, 10))
sns.heatmap(
    matrice_corr, 
    annot=True,          # Affiche les valeurs numériques dans les cases
    fmt=".2f",           # Arrondi à 2 décimales
    cmap="coolwarm",     # Couleur bleu (négatif) à rouge (positif)
    vmin=-1, vmax=1,     # Échelle bloquée entre -1 et 1
    linewidths=0.5
)
plt.title("Matrice de Corrélation : Profil, Revenus et Comportements d'Achat")
plt.show()
```

**Synthèse :**

* **Validation du biais de Récence :** La variable `Recency` affiche un coefficient de 0.00 avec l'intégralité des autres données (âge, revenus, dépenses). Cette absence totale de corrélation confirme mathématiquement que la récence a été lissée de manière artificielle à la source (exclusion des clients inactifs de plus de 100 jours) et n'apportera aucun pouvoir discriminant à l'algorithme de clustering.
* **Le revenu comme moteur de valeur :** La colonne `Income` présente une forte corrélation positive linéaire avec les rayons premiums comme le vin (0.58) et la viande (0.58). Le pouvoir d'achat s'impose comme le déclencheur principal du volume d'affaires global et dicte également une plus forte fréquentation des magasins physiques (0.53).
* **L'impact divergent de la structure familiale :** Les enfants en bas âge (`Kidhome`) et les adolescents (`Teenhome`) dictent des comportements financiers radicalement opposés. La présence de jeunes enfants fait chuter les budgets plaisir (-0.50 sur le vin et -0.44 sur la viande) au profit d'arbitrages de première nécessité. À l'inverse, la présence d'adolescents n'impacte pas ces rayons mais oriente fortement le foyer vers une stratégie de chasse aux prix (corrélation de 0.39 avec le nombre de promotions utilisées).
* **Légitimité de l'approche macro-comportementale :** Les dépenses des différents rayons sont fortement et positivement corrélées entre elles (ex: vin et viande à 0.56, viande et fruits à 0.54). Les clients consomment par « panier global ». Cela valide techniquement notre choix d'agréger ces lignes pour créer un score unique de dépense totale, réduisant ainsi la dimensionnalité sans perte d'information.

**Signaux faibles :**

* **Le rayon Or, un produit « Web-First » :** C'est le seul produit du dataset plus corrélé aux achats en ligne (0.42) qu'en magasin (0.38), révélant une dynamique d'achat d'impulsion ou de cadeaux stimulée spécifiquement par le canal digital.
* **Le Vin et le Magasin, le couple sacré :** La corrélation entre les dépenses de vin et les achats physiques est la plus élevée de toute la matrice (0.64), prouvant que le point de vente physique reste le pilier incontournable et l'épicentre de l'expérience client pour ce produit.
* **Le contre-pied de l'âge et du numérique :** L'âge présente une corrélation légèrement plus forte avec les achats web (0.15) qu'avec le réseau physique (0.13), brisant l'idée reçue d'une clientèle senior déconnectée ou réfractaire aux parcours omnicanaux.

**Relations croisées pour le ciblage comportemental :**

* **Le Web comme vecteur de promotions :** Le nombre d'achats en promotion est positivement lié aux achats sur internet (0.23) mais indépendant des achats en magasin (0.07). Les comportements opportunistes (chasse aux prix) sont donc digitalisés.
* **L'immunité promotionnelle des hauts revenus :** La corrélation négative entre `Income` et `NumDealsPurchases` (-0.08) démontre que la clientèle aisée est insensible aux mécaniques promotionnelles, achetant principalement au prix fort.
* **Le paradoxe budgétaire des adolescents :** La présence d'adolescents fait baisser le budget global alloué à la viande (-0.26). Cela indique une transition d'un panier "gourmet" (pièces nobles) vers un panier "volume/économique" (achats de masse), une distinction cruciale pour segmenter les profils familiaux.
* **La réalité de l'omnicanalité :** La forte corrélation entre les achats web et les achats en magasin (0.50) prouve que les meilleurs clients de l'enseigne naviguent activement entre les deux canaux, validant l'existence d'un profil de consommateur hybride.

```python
# ==========================================
# REQUÊTE SQL DE TRANSFORMATION (ELT)
# ==========================================
# 1. On calcule la médiane dans une CTE (Common Table Expression)
# 2. On applique le Feature Engineering (Âge, Ancienneté, Totaux)
# 3. On nettoie les catégories (CASE WHEN)
# 4. On filtre les outliers dans le WHERE
# 5. On exclut 'recency', 'dt_customer' et 'year_birth' (non sélectionnés)

SQL_TRANSFORM = """
    DROP TABLE IF EXISTS clients_transformed;

    CREATE TABLE clients_transformed AS
    WITH median_calc AS (
        SELECT PERCENTILE_CONT(0.5) WITHIN GROUP (ORDER BY income) as med_income
        FROM clients
    )
    SELECT 
        id,
        (2014 - year_birth) AS age,
        (DATE '2014-06-30' - dt_customer) AS anciennete_jours,
        CASE 
            WHEN education = '2n Cycle' THEN 'Master' 
            ELSE education 
        END AS education,
        CASE 
            WHEN marital_status IN ('Alone', 'YOLO') THEN 'Single' 
            ELSE marital_status 
        END AS marital_status,
        COALESCE(income, (SELECT med_income FROM median_calc)) AS income,
        kidhome, 
        teenhome,
        (kidhome + teenhome) AS children_total,
        mntwines, mntfruits, mntmeatproducts, mntfishproducts, mntsweetproducts, mntgoldprods,
        (mntwines + mntfruits + mntmeatproducts + mntfishproducts + mntsweetproducts + mntgoldprods) AS mnt_total,
        numdealspurchases, numwebpurchases, numcatalogpurchases, numstorepurchases, numwebvisitsmonth,
        (numwebpurchases + numcatalogpurchases + numstorepurchases) AS num_total_purchases,
        acceptedcmp3, acceptedcmp4, acceptedcmp5, acceptedcmp1, acceptedcmp2, 
        complain, response
    FROM clients
    WHERE (2014 - year_birth) <= 100 
      AND (income <= 120000 OR income IS NULL)
      AND marital_status != 'Absurd';
"""

# Exécution de la transformation en base de données
with psycopg2.connect(**CONN_PARAMS) as conn:
    with conn.cursor() as cur:
        print("Lancement de la transformation en base de données...")
        cur.execute(SQL_TRANSFORM)
        conn.commit()
        
        # Vérification du résultat
        cur.execute("SELECT COUNT(*) FROM clients_transformed;")
        nb_lignes = cur.fetchone()[0]
        print(f" Transformation terminée ! La table 'clients_transformed' contient {nb_lignes} lignes.")

# Extraction optionnelle pour vérifier dans Pandas
df_transformed = pd.read_sql("SELECT * FROM clients_transformed;", conn)
df_transformed.head()
```
